```python
graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 3), ('E', 6)],
    'C': [('F', 5)],
    'D': [('G', 2)],
    'E': [('G', 2)],
    'F': [('G', 1)],
    'G': []
}

# Heuristic values
heuristic = {
    'A': 7,
    'B': 6,
    'C': 5,
    'D': 4,
    'E': 3,
    'F': 2,
    'G': 0
}

def astar(start, goal):
    # open_list holds tuples of (node, g_cost)
    open_list = [(start, 0)]
    came_from = {}
    g_cost = {start: 0}

    while open_list:
        # Select node with minimum f = g + h
        current = min(open_list, key=lambda x: x[1] + heuristic[x[0]])
        open_list.remove(current)
        current_node = current[0]

        if current_node == goal:
            path = []
            curr = goal
            while curr in came_from:
                path.append(curr)
                curr = came_from[curr]
            path.append(start)
            path.reverse()
            return path, g_cost[goal]

        for neighbour, cost in graph[current_node]:
            new_cost = g_cost[current_node] + cost
            if neighbour not in g_cost or new_cost < g_cost[neighbour]:
                g_cost[neighbour] = new_cost
                came_from[neighbour] = current_node
                # If neighbour is already in open_list with higher cost, we update it
                open_list = [item for item in open_list if item[0] != neighbour]
                open_list.append((neighbour, new_cost))

    return None, float('inf')

# Main Program
start_node = input("Enter the start node: ").upper()
goal_node = input("Enter the goal node: ").upper()

path, cost = astar(start_node, goal_node)
if path:
    print("\nShortest path:", "->".join(path))
    print("Total cost:", cost)
else:
    print("Path not found")
```
